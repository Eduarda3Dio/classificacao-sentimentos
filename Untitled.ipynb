```python
# Manipulação de dados e visualização
import re
import pandas as pd
import numpy as np
import unicodedata
import seaborn as sns
import matplotlib.pyplot as plt

# Pré-Processamento e Machine Learning
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import joblib
```

### Limpeza dos textos
limpar os dados sujos

```python
# Remover linhas com valores ausentes
avaliacoes.dropna(subset = ['texto_review'], inplace = True)

print(avaliacoes.isnull().sum())
```

```python
# Função de limpeza de texto 
def texto_limpo(texto):
    # aqui tudo que nn for texto  ele retorna vazio
    if not isinstance(texto, str):
        return ""

    # Normaliza para a forma 'NFKD' que separa o caractere da acentuação
    # e depois remove os acentos (Nonspacing Mark)
    texto_sem_acentos = ''.join(c for c in unicodedata.normalize('NFKD', texto) if unicodedata.category(c) != 'Mn')

    # Converter para minúsculas
    texto_limpo = texto_sem_acentos.lower()

    # encontre tudo que não seja letra de a-z nem espaço e substitua por nada.
    texto_limpo = re.sub(r'[^a-z\s]', '', texto_limpo)

     # Remover espaços extras
    texto_limpo = re.sub(r'\s+', ' ', texto_limpo).strip()#remove espaços do inicio e final
    
    return texto_limpo
```

```python
# Aplica toda a função anterior de limpeza
# O lado direito executa a transformação. O lado esquerdo indica onde o resultado será armazenado
avaliacoes['texto_limpo'] = avaliacoes['texto_review'].apply(texto_limpo)
```

### Engenharia de Atributos

É a etapa onde transformamos dados brutos em "features" (atributos) úteis para o modelo

```python
#mapear em 0 e 1 os positivos e negativos
#o .map aplica algo aquela coluna
avaliacoes['sentimento_label'] = avaliacoes['sentimento'].map({'positivo': 1, 'negativo': 0})
```

### Dividir os dados entre treino e teste

```python
# Definir variáveis X (entrada) e y (saída)
X = avaliacoes['texto_limpo'] # textos q usara para aprender os padroes
y = avaliacoes['sentimento_label'] # os rotulos(0 e 1) que ele tentara aprender
```

```python
# Dividir os dados em treino e teste
# 0.25 = 25% para teste
#stratify=y	Preserva aproximadamente a proporção de sentimentos em treino e teste.
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size = 0.25, random_state = 42, stratify = y)
```

```python
# Pipeline 
pipeline = Pipeline([

    #transforma texto em valor numerico ignorando certas palavras
    ('tfidf', TfidfVectorizer(stop_words = ['de', 'a', 'o', 'que', 'e', 'do', 'da', 'em', 'um'])),

    # ajusta a escala dos valores numericos
    #('scaler', StandardScaler(with_mean = False)),

    # aqui manda para a regressao logistica ele pode estimar uma probabilidade de 0,85 para a classe 
    #positiva e, pelo limiar de decisão, prever 1.
    ('logreg', LogisticRegression(solver = 'liblinear', random_state = 42, max_iter = 1000)) 
])
```

```python
type(pipeline)
```

```python
C: controla a força da regularização.
penalty: define como os pesos são penalizados.
max_iter: limita o trabalho do algoritmo de otimização.
GridSearchCV: compara configurações usando validação cruzada.
```

```python
# Definir o grid de hiperparâmetros para otimização(onde escolhemos oq sera testado)
parametros_grid = {
    'tfidf__max_features': [500, 1000, 2000], #Limitar o vocabulário a 500, 1.000 ou 2.000 características
    'tfidf__ngram_range': [(1, 1), (1, 2)], #Usar palavras isoladas ou combinações de até duas palavras
    'logreg__C': [0.1, 1, 10], #Controlar a intensidade da regularização. Valores menores significam regularização mais forte
    'logreg__penalty': ['l1', 'l2'], #Comparar regularização L1 e L2.
    'logreg__max_iter': [5000, 6000] #Definir o limite de iterações do algoritmo de otimização.
}
```

```python
# Configurar o GridSearchCV(onde busca diferentes combinaçoes de hiperparametros)
grid_search = GridSearchCV(
    pipeline,              # conjunto de transformações e modelo que será avaliado
    parametros_grid,       # combinações de configurações/ hiperparâmetros que serão testadas 
    cv = 5,                # divide os dados em 5 para validação cruzada (5-fold cross-validation)
    n_jobs = -1,           # Usa todos os núcleos disponíveis do processador para acelerar o processo
    scoring = 'accuracy',  # Métrica usada para avaliar o desempenho de cada combinação (aqui, acurácia)
    verbose = 1            # exibe informações sobre o andamento da execução. (1 exibe progresso básico)
)
```

```python
print("\nIniciando o treinamento do modelo com otimização de hiperparâmetros...\n")
grid_search.fit(X_treino, y_treino)
```

```python
grid_search.best_score_

#1.0 significa uma acuracia de 100%
```

```python
#mostra oq o modelo escolheu
grid_search.best_params_
```

Regularização (C)
0,1 Entre os valores testados, foi escolhida a regularização mais forte.

Penalidade
L2 Penaliza pesos elevados sem, em geral, zerar os coeficientes.

Número máximo de características
500 O vocabulário foi limitado a 500 características.

N-gramas
(1, 1) O modelo utiliza palavras individuais, sem combinações de duas palavras.

Iterações máximas
5.000 Foi o limite escolhido entre os valores testados; isso não significa que o algoritmo precisou executar todas as iterações.

```python
# Obter o melhor modelo recupera o pipeline completo com os hiperparâmetros selecionados pelo GridSearchCV
melhor_modelo = grid_search.best_estimator_
type(melhor_modelo)
```

```python
# Previsões no conjunto de teste utiliza o melhor pepiline para gerar as previsões das avaliações que ficaram separadas para teste
y_pred = melhor_modelo.predict(X_teste)
```

```python
# Calcular as métricas de avaliação
acuracia = accuracy_score(y_teste, y_pred) #proporção de previsões corretas no conjunto de teste
report = classification_report(y_teste, y_pred, target_names = ['Negativo', 'Positivo'])
```

```python
print(f"\nAcurácia do Modelo: {acuracia:.2%}\n")
print("Relatório de Classificação:\n")
print(report) #relatorio das acuracias
```
